```python
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
```

#### Train_test_split

```python
X_train, X_test, y_train,y_test = train_test_split(df.iloc[:,1:],df.iloc[:,0],test_size=0.2)
```

```python
X_train.shape
```

```python
y_test # Target: M - Malignant B - Benign
```

#### Scaling

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
X_train
```

```python
y_train
```

#### Label Encoding
* m aur b humera model nahi samajh sakta hai isiliya hum label incoding ker reha hai

```python
encoder = LabelEncoder()
y_train = encoder.fit_transform(y_train)
y_test = encoder.transform(y_test)
```

```python
y_train
```

### Numpy array to pytorch tensor

```python
X_train_tensor = torch.from_numpy(X_train).float()
X_test_tensor = torch.from_numpy(X_test).float()

y_train_tensor = torch.from_numpy(y_train).float().view(-1, 1)
y_test_tensor = torch.from_numpy(y_test).float().view(-1, 1)
```

```python
X_train_tensor.dtype
```

```python
X_train_tensor.shape
```

##### conclusion --> our deep learning model work on tensor

```python
# our model have 30 weights and 1 bias
class MySimpleNN(nn.Module):
    def __init__(self, num_features):
        super().__init__()
        self.linear = nn.Linear(num_features, 1)
        self.sigmoid = nn.Sigmoid()

    def forward(self, features):
        out = self.linear(features)
        out = self.sigmoid(out)

        return out
```

#### Important Parameters

```python
learning_rate = 0.1
epochs = 25
```

```python
# Define loss function
loss_function = nn.BCELoss()
```

#### Training pipeline

```python
# crate model
model = MySimpleNN(X_train_tensor.shape[1])

# define optimizer
optimizer = torch.optim.SGD(model.parameters(), lr=learning_rate)

# define loop
for epoch in range(epochs):
    
    # forward pass
    y_pred = model(X_train_tensor)
    # loss calculate
    loss = loss_function(y_pred, y_train_tensor)
    # clear gradients
    optimizer.zero_grad()
    # backword pass
    loss.backward()
    # parameter updations
    optimizer.step()

    # print loss in each epoch
    print(f'Epoch: {epoch + 1}, loss: {loss.item()}')
```

* grad.zero_() is an in-place PyTorch method that resets the accumulated gradients of a specific tensor to zero, ensuring that subsequent backward() passes compute fresh gradients for the current batch rather than adding to previous values.

```python
model.linear.bias
```

#### Evaluation

```python
# model evaluation
with torch.no_grad(): # no_grad mean hum ko abhi koi backword propagation nahi kerna hai isiliya hum no grad ker reha hai
    y_pred = model(X_test_tensor)
    y_pred = (y_pred > 0.5).float() # means less than 0.5 --> and greater then 0.5 -> 1
    accuracy = (y_pred == y_test_tensor).float().mean()
    print(f"Accuracy: {accuracy.item()}")
```
