### Creating a tensor

```python
# check type
type(a)
```

```python
# using zeros
torch.zeros(2,3)
```

```python
# using ones
torch.ones(2,3)
```

```python
# using rand
torch.rand(2,3)
```

```python
torch.rand(2,3) # our above is not match so we use manual_seed
```

```python
# manual seed for each time randome value is same
torch.manual_seed(100)
torch.rand(2,3)
```

```python
torch.manual_seed(100)
torch.rand(2,3) # above value is same
```

```python
# using tensor
torch.tensor([[1,2,3,4],[5,4,3,2]])
```

```python
# Other ways to create tensor

# Arange
print("using arange -->", torch.arange(0,10,2))

# Using Linespace
print("Using linespace -->", torch.linspace(0,10,10))

# Using eye
print("using eye -->", torch.eye((5))) # digonal element

# Using full
print("using full -->", torch.full((3,3),5))
```

### Tensor Shape

```python
x = torch.tensor([[1,2,3,4],[4,3,2,1]])
```

```python
x.shape
```

```python
# If we want to copy shape of any other tensor 
torch.empty_like(x)
```

```python
torch.zeros_like(x)
```

```python
torch.ones_like(x)
```

```python
torch.rand_like(x,dtype=torch.float64)
```

### Tensor Data Types

```python
# find data type
x.dtype
```

```python
# assing data type
torch.tensor([1.0,2.0,3.0], dtype=torch.int32)
```

```python
torch.tensor([1,2,3,4],dtype=torch.float64)
```

```python
# using to() to change datatype of existing datatype
x.to(torch.float64)
```
