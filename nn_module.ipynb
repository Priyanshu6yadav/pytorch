### Single neural network

```python
# Create a model class
import torch
import torch.nn as nn

class Model(nn.Module):
    def __init__(self, num_features):

        super().__init__()
        self.linear = nn.Linear(num_features, 1) # num_feature --> input and 1 --> output
        self.sigmoid = nn.Sigmoid()

    def forward(self, features):

        out = self.linear(features) # z = w*x + b will be done 
        out = self.sigmoid(out) # we find sigmoid  means y_pred

        return out
```

```python
features.shape[1]
```

```python
# Create datasets
features = torch.rand(10,5) # rows --> 10 column --> 5

# create model
model = Model(features.shape[1]) # feature.shape[1] --> means first row shape --> 5

# call model for forward pass
# model.forward(featues)
model(features) # --> y_pred
```

```python
# show model weights
model.linear.weight
```

```python
# show model bias
model.linear.bias
```

```python
from torchinfo import summary

summary(model, input_size=(10,5))
```

* conclusion 6 trainable parameter 5 weight and 1 bias --> 6

#### nn with hidden neural network 
* 5 --> input and 3 hidden neuron and 1 output

* 5 * 3 --> 15 weight and 3 * 1 --> 3 weight and 3 + 1 four bias , Total trainable parameter --> 22
* we use relu in hidden layer and sigmoid in output

```python
class Model(nn.Module):
    def __init__(self, num_features):

        super().__init__()
        self.linear1 = nn.Linear(num_features, 3) # HIDDEN LAYER output may hidden layer sa 3 weights nikal reha hai
        self.relu = nn.ReLU()
        self.linear2 = nn.Linear(3,1) # LAST NEURON
        self.sigmoid = nn.Sigmoid()
    def forward(self, features):

        out = self.linear1(features)
        out = self.relu(out)
        out = self.linear2(out)
        out = self.sigmoid(out) 

        return out
```

```python
# Create datasets
features = torch.rand(10,5) # rows --> 10 column --> 5

# create model
model = Model(features.shape[1]) # feature.shape[1] --> means first row shape --> 5

# call model for forward pass
# model.forward(featues)
model(features) # --> y_pred
```

```python
model.linear1.weight # --> 15 weights
```

```python
model.linear2.weight # --> 3 weights
```

```python
from torchinfo import summary
summary(model, input_size=(10,5))
```

#### Using sequential container

```python
class Model(nn.Module):
    def __init__(self, num_features):

        super().__init__()
        self.network = nn.Sequential(
        nn.Linear(num_features, 3), # HIDDEN LAYER output may hidden layer sa 3 weights nikal reha hain
        nn.ReLU(),
        nn.Linear(3,1), # LAST NEURON
        nn.Sigmoid(),
        )
    def forward(self, features):
        out = self.network(features)
        return out
```
