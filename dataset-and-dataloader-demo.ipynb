```python
from sklearn.datasets import make_classification
import torch
```

```python
# Step 1: Create a synthetic classification dataset using sklearn
X,y = make_classification(
    n_samples=10, # Number of sample
    n_features=2, # number of features
    n_informative=2, # Number of informative features
    n_redundant=0, # Number of redundant features
    n_classes=2, # Number of classes
    random_state=42 # For reproducibility
)
```

```python
# convert the data to Pytorch Tensors
X = torch.tensor(X, dtype=torch.float32)
y = torch.tensor(y, dtype=torch.long)
```

```python
from torch.utils.data import Dataset, DataLoader
```

```python
class CustomDataset(Dataset):
    def __init__(self, features, labels):
        
        self.features = features
        self.labels = labels

    def __len__(self):
        return self.features.shape[0] # reture number of features

    def __getitem__(self, index):
        return self.features[index], self.labels[index]
```

```python
dataset = CustomDataset(X, y)
```

```python
len(dataset)
```

```python
dataset[3] # --> third row
```

```python
dataloader = DataLoader(dataset, batch_size=2, shuffle=True) # batch_size =2 --> [tow columns] and shuffle = true means shuffle all columns
```

* num_workers=4  --> Uses 4 subprocesses for data loading

```python
for batch_features, batch_labels in dataloader:
    print(batch_features)       # shape: torch.Size([2, 2]) -> 2 rows, 2 features
    print(batch_labels)         # shape: torch.Size([2])    -> 2 labels
    print("-" * 50)
```
